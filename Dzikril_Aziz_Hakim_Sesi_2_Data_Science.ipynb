```python
modus = df["Skor_Kinerja"].mode()[0]

print("Modus Skor Kinerja:", modus)
```

```python
varians = df["Skor_Kinerja"].var()

print("Varians Skor Kinerja:", varians)
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Skor_Kinerja"], bins=10, edgecolor="black")

plt.title("Distribusi Skor Kinerja Karyawan")
plt.xlabel("Skor Kinerja")
plt.ylabel("Jumlah Karyawan")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Departemen",
    y="Skor_Kinerja"
)

plt.title("Boxplot Skor Kinerja per Departemen")
plt.xlabel("Departemen")
plt.ylabel("Skor Kinerja")

plt.show()
```

```python
rata_lama_bekerja = df.groupby("Departemen")["Lama_Bekerja_Tahun"].mean()

plt.figure(figsize=(10, 6))

rata_lama_bekerja.plot(kind="bar")

plt.title("Rata-rata Lama Bekerja per Departemen")
plt.xlabel("Departemen")
plt.ylabel("Rata-rata Lama Bekerja (Tahun)")

plt.xticks(rotation=0)

plt.show()
```

```python
rata_lama_bekerja = df.groupby("Departemen")["Lama_Bekerja_Tahun"].mean()

print("Rata-rata Lama Bekerja per Departemen:")
print(rata_lama_bekerja)
```
